```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Dataset load karein
df = pd.read_excel('global_superstore_2016 avi.xlsx', sheet_name='Orders')

# 1. Total Revenue by Category
revenue_by_cat = df.groupby('Category')['Sales'].sum().reset_index()
print("--- Revenue by Category ---")
print(revenue_by_cat)

# 2. Bar Chart generate karna
plt.figure(figsize=(8, 5))
sns.barplot(data=revenue_by_cat, x='Category', y='Sales', palette='Blues_d')
plt.title('Total Revenue by Product Category')
plt.xlabel('Category')
plt.ylabel('Total Sales ($)')
plt.savefig('revenue_by_category.png')
plt.show()

# 3. Profit by Region
profit_by_region = df.groupby('Region')['Profit'].sum().reset_index().sort_values(by='Profit', ascending=False)
print("\n--- Top Regions by Profit ---")
print(profit_by_region.head())
```

Executive Summary & Final Report:
- Project Overview: Conducted an end-to-end data analysis on the Global Superstore dataset to evaluate sales performance, product categories, and regional profitability.
- Key Findings: Technology generates the highest total sales, indicating strong demand for electronic and communication tools. However, discounts and shipping costs impact net profit margins in certain regions.
- Recommendations: Focus marketing efforts on high-margin Technology products, implement discount caps in low-profit areas, and optimize shipping for bulky items.
