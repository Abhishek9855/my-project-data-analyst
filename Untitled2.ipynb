```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('global_superstore_2016 avi.xlsx', sheet_name='Orders')

print("--- 1. Missing Values Before Cleaning ---")
print(df.isnull().sum())
```

```python
df['Sales'] = df['Sales'].fillna(df['Sales'].median())
df['Profit'] = df['Profit'].fillna(df['Profit'].median())
```

```python
filtered_df = df[df['Category'] == 'Technology']
print(f"\nTotal Technology rows: {len(filtered_df)}")
```

```python
df['Total_Cost'] = df['Sales'] - df['Profit']
df['Profit_Margin'] = (df['Profit'] / df['Sales']) * 100

print("\n--- New Columns Created Successfully ---")
print(df[['Customer Name', 'Category', 'Sales', 'Profit', 'Profit_Margin']].head())
```
