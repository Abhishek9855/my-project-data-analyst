```python
import pandas as pd

# Excel file aur Orders sheet load karne ke liye
df = pd.read_excel('global_superstore_2016 avi.xlsx', sheet_name='Orders')

print("--- First 5 Rows ---")
print(df.head())

print("\n--- Basic Info ---")
print(df.info())
```

```python
df = df.dropna(subset=['Sales', 'Category', 'Customer Name'])
df = df.drop_duplicates()
print(f"Total rows after cleaning: {len(df)}")
```

```python
revenue_by_category = df.groupby('Category')['Sales'].sum().reset_index()
print("\n--- Total Revenue by Category ---")
print(revenue_by_category)
```

```python
sorted_df = df.sort_values(by=['Category', 'Sales'], ascending=[True, False])
print("\n--- Sorted Data (Top Rows) ---")
print(sorted_df[['Category', 'Customer Name', 'Sales']].head())
```

```python
correlation_matrix = df.select_dtypes(include=['number']).corr()
print("\n--- Correlation Matrix ---")
print(correlation_matrix)
```
